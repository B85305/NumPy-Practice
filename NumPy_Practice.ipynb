```python
Perform matrix multiplication on [[1, 2], [3, 4]] and [[5, 6], [7, 8]].
```

```python
Given [[10, 20, 30], [40, 50, 60], [70, 80, 90]], find the row sums, column sums, overall mean, and elements greater than 45.
```

```python
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print("Row Sums:", np.sum(arr, axis=1))
print("Column Sums:", np.sum(arr, axis=0))
print("Overall Mean:", np.mean(arr))
print("Elements Greater than 45:", arr[arr > 45])
```

```python
Given two NumPy arrays, find the elements that are common to both arrays.
```

```python

array1 = np.array([10, 20, 30, 40, 50])
array2 = np.array([30, 40, 50, 60, 70])

common_elements = np.intersect1d(array1, array2)

print(common_elements)
```
